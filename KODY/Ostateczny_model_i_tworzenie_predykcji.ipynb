# Wczytywanie danych do projektu

```python
import pandas as pd

# Wczytywanie danych treningowych
train_data = pd.read_csv('artifical_train_data.csv')

# Wczytywanie etykiet dla danych treningowych
train_labels = pd.read_csv('artifical_train_labels.csv')

# Wczytywanie danych testowych
test_data = pd.read_csv('artifical_test_data.csv')

# Wyświetlanie pierwszych kilku wierszy, aby zweryfikować wczytanie
print(" Dane treningowe:")
print(train_data.head())
print("Etykiety treningowe:")
print(train_labels.head())
print("Dane testowe:")
print(test_data.head())
```

```python
print("Informacje o danych treningowych:")
train_data.info()
```

# Pipeline do trenowania i predykcji

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectFromModel
from sklearn.model_selection import KFold, cross_val_score,RepeatedStratifiedKFold
import numpy as np
```

```python
from imblearn.pipeline import Pipeline as ImbPipeline
from imblearn.over_sampling import SMOTE
from sklearn.neighbors import KNeighborsClassifier
from sklearn.decomposition import PCA
BEST_PARAMS = {
    'n_neighbors':5,
    'weights': 'distance'
}
clf_for_selection = RandomForestClassifier(random_state=42)
pipeline = ImbPipeline([
    ('Scaler', StandardScaler()),
    ('Selector', SelectFromModel(clf_for_selection, threshold="1.4*mean")),
    #('PCA', PCA(n_components=0.975)), # n_components=0.95 wybierze tyle składowych, by zachować 95% informacji
    ('SMOTE', SMOTE(random_state=42)),
    ('Classifier', KNeighborsClassifier(**BEST_PARAMS))
])
```

# Ocena modelu za pomocą 3-krotnej kroswalidacji

```python
# Przygotowanie do kroswalidacji
kfold = RepeatedStratifiedKFold(n_splits=10,n_repeats=20)
scores = cross_val_score(pipeline, train_data, train_labels.values.ravel(), cv=kfold, scoring='balanced_accuracy')

print(f"Wyniki kroswalidacji (-krotnej) z metryką balanced_accuracy:")
print(scores)
print(f"Średnia dokładność (balanced_accuracy): {scores.mean():.2f}")
```

```python
# Trenowanie modelu
# Używamy .values.ravel() do przekształcenia etykiet na odpowiedni format
print("Rozpoczynam trenowanie modelu...")
pipeline.fit(train_data, train_labels.values.ravel())
print("Model został wytrenowany.")
```

```python
# Predykcja na danych testowych
print("Rozpoczynam predykcję na danych testowych...")
predictions = pipeline.predict_proba(test_data)[:, 1]
print("Predykcja została zakończona.")

# Wyświetlanie przykładowych predykcji
print("Przykładowe predykcje (prawdopodobieństwo klasy 1):")
print(predictions[:10])
```

```python
# Zapisywanie predykcji do pliku TXT
file_name = '333076_artifical_prediction.txt'
np.savetxt(file_name, predictions, fmt='%.18f')
print(f"Predykcje zostały zapisane do pliku '{file_name}'.")
```
