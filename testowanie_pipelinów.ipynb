Pierwszy test-testowanie wielu modeli i skalarów aby sprawdzić który najlepszy (UWAGA ODPALANIE WSZYSTKICH NARAZ BĘDZIE LAGOWAĆ DŁUGO )(Dlatego zahashowałem część z nich)

```python
# NOWA SEKCJA: BARDZO OBSZERNE TESTY MODELI


# Wczytywanie danych
train_data = pd.read_csv('artifical_train_data.csv')
train_labels = pd.read_csv('artifical_train_labels.csv')
test_data = pd.read_csv('artifical_test_data.csv')

# Przygotowanie danych
X = train_data
y = train_labels.values.ravel()

# Definiowanie list komponentów do przetestowania
scalers = [
    ('StandardScaler', StandardScaler()),
    #('MinMaxScaler', MinMaxScaler()),
    ('RobustScaler', RobustScaler()),
    ('Normalizer', Normalizer())
]

feature_selectors = [
    ('SelectKBest_f_classif', SelectKBest(f_classif)),
    #('SelectKBest_mutual_info', SelectKBest(mutual_info_classif)),
    ('SelectFromModel_RF', SelectFromModel(RandomForestClassifier(random_state=42)))
]

classifiers = [
    #('LogisticRegression', LogisticRegression(random_state=42)),
    ('RandomForestClassifier', RandomForestClassifier(random_state=42)),
    ('GradientBoostingClassifier', GradientBoostingClassifier(random_state=42)),
    #('SVC', SVC(probability=True, random_state=42)),
    ('KNeighborsClassifier', KNeighborsClassifier()),
    #('DecisionTreeClassifier', DecisionTreeClassifier(random_state=42)),
    #('GaussianNB', GaussianNB()),
    #('LinearDiscriminantAnalysis', LinearDiscriminantAnalysis()),
    #('QuadraticDiscriminantAnalysis', QuadraticDiscriminantAnalysis())
]

# Siatki parametrów dla każdego klasyfikatora
param_grids = {
    # 'LogisticRegression': {
    #     'Classifier__C': [0.1, 1, 10],
    #     'Classifier__penalty': ['l1', 'l2'],
    #     'Classifier__solver': ['liblinear'] # liblinear supports both l1 and l2
    # },
    'RandomForestClassifier': {
        'Classifier__n_estimators': [70],
        'Classifier__max_depth': [None],
        'Classifier__min_samples_split': [2, 5]
    },
    'GradientBoostingClassifier': {
        'Classifier__n_estimators': [50, 100],
        'Classifier__learning_rate': [0.05, 0.1, 0.2],
        'Classifier__max_depth': [3, 5, 7]
    },
    'SVC': {
        'Classifier__C': [100, 10],
        'Classifier__gamma': ['scale'],
        'Classifier__kernel': ['rbf', 'linear']
    },
    'KNeighborsClassifier': {
        'Classifier__n_neighbors': [3, 5, 7],
        'Classifier__weights': ['uniform', 'distance']
    },
    # 'DecisionTreeClassifier': {
    #     'Classifier__max_depth': [None, 10, 20],
    #     'Classifier__min_samples_split': [2, 5]
    # },
    #'GaussianNB': {},
    #'LinearDiscriminantAnalysis': {
    #    'Classifier__solver': ['svd', 'lsqr', 'eigen']
    #},
    #'QuadraticDiscriminantAnalysis': {}
}

# Parametry dla selekcji cech
feature_selector_params = {
    'SelectKBest_f_classif': {
        'Selector__k': [10, 20, 'all']
    },
    'SelectKBest_mutual_info': {
        'Selector__k': [10, 20, 'all']
    },
    'SelectFromModel_RF': {
        'Selector__threshold': ['mean', 'median', '1.25*mean']
    }
}


# Uruchomienie pętli testującej
kfold = KFold(n_splits=3, shuffle=True, random_state=42)
results = []

for scaler_name, scaler in scalers:
    for selector_name, selector in feature_selectors:
        for clf_name, classifier in classifiers:
            
            pipeline = Pipeline([
                ('Scaler', scaler),
                ('Selector', selector),
                ('Classifier', classifier)
            ])
            
            # Połączenie siatek parametrów
            param_grid = {}
            if clf_name in param_grids:
                param_grid.update(param_grids[clf_name])
            if selector_name in feature_selector_params:
                param_grid.update(feature_selector_params[selector_name])

            # Pomijanie niekompatybilnych kombinacji
            if not param_grid and not (clf_name == 'GaussianNB' or clf_name == 'QuadraticDiscriminantAnalysis'):
                 # QDA and GNB might not have params, but others should
                continue

            print(f"--- Testing Pipeline: {scaler_name} -> {selector_name} -> {clf_name} ---")

            # GridSearchCV
            grid_search = GridSearchCV(pipeline, param_grid, cv=kfold, scoring='balanced_accuracy', verbose=1, n_jobs=-1, error_score='raise')
            
            try:
                grid_search.fit(X, y)
                
                best_score = grid_search.best_score_
                best_params = grid_search.best_params_
                
                print(f"Best Score: {best_score}")
                print(f"Best Params: {best_params}")
                
                results.append({
                    'scaler': scaler_name,
                    'selector': selector_name,
                    'classifier': clf_name,
                    'best_score': best_score,
                    'best_params': best_params
                })
            except Exception as e:
                print(f"Error during GridSearchCV for {scaler_name} -> {selector_name} -> {clf_name}: {e}")
            
            print("\\n" + "="*80 + "\\n")

# Wyświetlenie podsumowania wyników
print("\\n\\n--- All Results Summary ---")
results_df = pd.DataFrame(results)
results_df = results_df.sort_values(by='best_score', ascending=False)
print(results_df)

# Zapisanie wyników do pliku
results_df.to_csv('extensive_grid_search_results.csv', index=False)
print("\\nResults saved to 'extensive_grid_search_results.csv'")
```

```python
# Definiowanie potoku do tuningu
clf_for_selection = RandomForestClassifier(random_state=42)
pipeline = Pipeline([
    ('Scaler', StandardScaler()),
    ('Selector', SelectFromModel(clf_for_selection)),
    ('Classifier', RandomForestClassifier(random_state=42))
])

# Definiowanie siatki parametrów
param_grid = {
    'Selector__threshold': ['mean'],# wcześniej testowałem inne ale usunąłem nie testowałem wszystkich naraz bo by to milion lat zajęło
    'Classifier__n_estimators': [70],
    'Classifier__max_depth': [None],
}

# Uruchomienie GridSearchCV
kfold = KFold(n_splits=5, shuffle=True, random_state=42)
grid_search = GridSearchCV(pipeline, param_grid, cv=kfold, scoring='balanced_accuracy', verbose=1, n_jobs=-1)
grid_search.fit(X, y)

# Wyświetlenie najlepszych parametrów i wyniku
print("Najlepsze parametry:", grid_search.best_params_)
print("Najlepszy wynik balanced_accuracy:", grid_search.best_score_)

```

testowanie Pipelinu  KNeighbour z PCA (bo wydawało mi się żę może podnieść skuteczność ale coś nie szło)

```python

# The pipeline from data_loading.ipynb
clf_for_selection = RandomForestClassifier(random_state=42)
pipeline = ImbPipeline([
    ('Scaler', StandardScaler()),
    ('Selector', SelectFromModel(clf_for_selection)),
    ('PCA', PCA()),
    ('SMOTE', SMOTE(random_state=42)),
    ('Classifier', KNeighborsClassifier())
])

# Definiowanie siatki parametrów
param_grid = {
    'PCA__n_components': [0.9, 0.95, 0.99],
    'Selector__threshold': ['1.25*mean', '1.4*mean', '1.5*mean'],
    'Classifier__n_neighbors': [3,5, 7]
}

# Uruchomienie GridSearchCV
kfold = RepeatedStratifiedKFold(n_splits=5,n_repeats=2)
grid_search = GridSearchCV(pipeline, param_grid, cv=kfold, scoring='balanced_accuracy', verbose=1, n_jobs=-1)
grid_search.fit(X, y)

# Wyświetlenie najlepszych parametrów i wyniku
print("Najlepsze parametry:", grid_search.best_params_)
print("Najlepszy wynik balanced_accuracy:", grid_search.best_score_)
```

Testowanie Pipelinu KNeighbour (Najlepszy model który potem używam już w ostatecznym kodzie)

```python


# 1. Konfiguracja walidacji
kfold = StratifiedKFold(n_splits=15, shuffle=True, random_state=42)

# 2. Parametry, które wybrałeś jako najlepsze
BEST_PARAMS = {
    'n_neighbors': 4,
    'weights': 'distance'
}
clf_for_selection = RandomForestClassifier(random_state=42)
pipeline = Pipeline([
    ('Scaler', StandardScaler()),
    ('Selector', SelectFromModel(clf_for_selection)),
    ('Classifier',KNeighborsClassifier())
])

# Definiowanie siatki parametrów
param_grid = {
    'Selector__threshold': ['1.4*mean',"1.25*mean",'1.35*mean'],# wcześniej testowałem inne ale usunąłem nie testowałem wszystkich naraz bo by to milion lat zajęło
    
}
# 3. Definicja potoku (Pipeline)
model_pipeline = ImbPipeline([
    ('Scaler', StandardScaler()),
    ('SMOTE', SMOTE(random_state=42)),
    ('Selector', SelectFromModel(clf_for_selection,
        threshold="1.25*mean"
    )),
    ('Classifier', KNeighborsClassifier(**BEST_PARAMS))
])

# --- NOWA SEKCJA: OBLICZANIE WYNIKU ---
print("Obliczam Balanced Accuracy metodą Cross-Validation... Proszę czekać.")
# cv_results zwróci tablicę 5 wyników (dla każdego folda)
cv_results = cross_val_score(model_pipeline, X, y, cv=kfold, scoring='balanced_accuracy', n_jobs=-1)

print("\n" + "="*35)
print(f"Średni Balanced Accuracy: {np.mean(cv_results):.4f} (+/- {np.std(cv_results):.4f})")
print(f"Wyniki poszczególnych foldów: {np.round(cv_results, 4)}")
print("="*35 + "\n")
```
