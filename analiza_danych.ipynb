# Analiza Danych

W tym notatniku dokonamy analizy danych w celu zrozumienia ich charakterystyk, co pomoże w doborze odpowiedniego modelu predykcyjnego.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Wczytanie danych
train_data = pd.read_csv('artifical_train_data.csv')
train_labels = pd.read_csv('artifical_train_labels.csv')
test_data = pd.read_csv('artifical_test_data.csv')

# Wyświetlenie pierwszych wierszy danych
print("Train Data Head:")
print(train_data.head())
print("
Train Labels Head:")
print(train_labels.head())
print("
Test Data Head:")
print(test_data.head())
```

```python
# Informacje o danych
print("
Train Data Info:")
train_data.info()

print("
Test Data Info:")
test_data.info()
```

```python
# Statystyki opisowe
print("
Train Data Description:")
print(train_data.describe())
```

```python
# Sprawdzenie brakujących wartości
print("
Missing values in Train Data:")
print(train_data.isnull().sum())
```

```python
# Analiza zmiennej docelowej
print("
Train Labels value counts:")
print(train_labels.value_counts())
```

```python
# Wizualizacja rozkładu zmiennej docelowej
sns.countplot(x=train_labels.columns[0], data=train_labels)
plt.title('Rozkład zmiennej docelowej')
plt.show()
```

```python
# Połączenie danych treningowych z etykietami dla dalszej analizy
train_full = pd.concat([train_data, train_labels], axis=1)

# Macierz korelacji
plt.figure(figsize=(12, 10))
sns.heatmap(train_full.corr(), fmt='.2f', cmap='coolwarm')
plt.title('Macierz korelacji')
plt.show()
```

```python
# Pairplot dla kilku wybranych cech
# Wybieramy kilka cech, aby uniknąć zbyt dużego wykresu
features_to_plot = list(train_data.columns[:4]) + [train_labels.columns[0]]
sns.pairplot(train_full[features_to_plot], hue=train_labels.columns[0])
plt.show()
```
